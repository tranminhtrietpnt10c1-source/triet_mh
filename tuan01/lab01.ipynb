#### Bài 3. Kiểm tra thông tin Tensor

```python
import torch
a = torch.zeros(2, 3, 4)
print("Tensor:")
print(a)
print("Shape:", a.shape)
print("Size:", a.size())
print("Number of dimensions:", a.ndim)
print("Data type:", a.dtype)
```

torch.Size([2, 3, 4]) có nghĩa là tensor có 3 chiều (3D):
2 là 2 khối ,
3 là mỗi khối 3 hàng,
4 là mỗi hàng 4 phần tử.

#### Bài 4. Slicing và Indexing


```python
import torch
a = torch.randn(3, 4, 5)
print("Original tensor:")
print(a)
print("\nTensor a[1]:")
print(a[1])
print("\nTensor a[1:, 2:4]:")
print(a[1:, 2:4])
print("\nTensor a[:,: ,-2:]:")
print(a[:,:,-2:])
```

a.shape=torch.Size([3,4,5])

a[1].shape=torch.Size([4,5])
4 hàng 
5 cột

a[1:, 2:4].shape = torch.Size([2, 2, 5])

1: lấy khối index 1 và 2 là 2 khối,
2:4 lấy hàng index 2 và 3 là 2 hàng,
Không cắt chiều cuối lấy đủ 5 cột

#### Bài 5. Reshape và Flatten

```python
import torch
a = torch.randn(3, 4)
print("Original:")
print(a)
print("\nFlatten:")
print(a.ravel())
print("\nReshape to 3 x 2 x 2:")
print(a.reshape(3, 2, 2))
```

số phần tử trước và sau reshape không thay đổi chỉ thay đổi chiều và vị trí sắp xếp
 

```python
import torch
a = torch.randn(3, 4)
print("Original:")
print(a)
print("\nFlatten:")
print(a.ravel())
print("\nReshape to 2 x 6:")
print(a.reshape(2,6))
```

#### Bài 6. Phép toán và nhân ma trận

```python
import torch
a = torch.randn(2, 3)
b = torch.randn(2, 3)
print("a:")
print(a)
print("b:")
print(b)
print("\na + b:")
print(a + b)
print("\na / b:")
print(a / b)
print("\na^2:")
print(a ** 2)
print("\na * b:")
print(a * b)
print("\nMatrix multiplication:")
print(a @ b.T)
```

##### a*b là nhân từng phần tử 
##### a @ b.T là nhân ma trận a vơi ma trận chuyển vị của b

##### a @ b.T = torch.Size([2,2])

#### Bài 7. Các hàm thống kê trên Tensor


```python
import torch
a = torch.randn(3, 4)
print(a)
print("\nMean:")
print(torch.mean(a, dim=0))
print("\nStandard deviation:")
print(torch.std(a, dim=0))
print("\nCumulative sum:")
print(torch.cumsum(a, dim=0))
```

thay dim=1

```python
import torch
a = torch.randn(3, 4)
print(a)
print("\nMean:")
print(torch.mean(a, dim=1))
print("\nStandard deviation:")
print(torch.std(a, dim=1))
print("\nCumulative sum:")
print(torch.cumsum(a, dim=1))
```

Mean và Standard deviation: giảm 1 phần tử so với dim =0

Cumulative sum: số lượng phần tử không đổi
